### 3.8 Adding columns for "Total Score" and "Average"

```python
df['total score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average'] = df['total score']/3
df.head()
```

```python
reading_full = df[df['reading_score'] == 100]['average'].count()
writing_full = df[df['writing_score'] == 100]['average'].count()
math_full = df[df['math_score'] == 100]['average'].count()

# we include 'average' feature to simply give us the count. 
# else it displays rows satisfying the condition in all features

print(f'Number of students with full marks in Maths: {math_full}')
print(f'Number of students with full marks in Writing: {writing_full}')
print(f'Number of students with full marks in Reading: {reading_full}')
```

```python
reading_less_20 = df[df['reading_score'] <= 20]['average'].count()
writing_less_20 = df[df['writing_score'] <= 20]['average'].count()
math_less_20 = df[df['math_score'] <= 20]['average'].count()

print(f'Number of students with less than 20 marks in Maths: {math_less_20}')
print(f'Number of students with less than 20 marks in Writing: {writing_less_20}')
print(f'Number of students with less than 20 marks in Reading: {reading_less_20}')
```

#####  Insights
 - From above values we get students have performed the worst in Maths 
 - Best performance is in reading section

#### 4.3 Multivariate analysis using pieplot

```python
plt.rcParams['figure.figsize'] = (15, 10)
'''
Sets the default size of every figure you draw afterward: 30 inches wide, 12 inches tall. 
rcParams is matplotlib global settings dictionary, so this applies to all later plots in the notebook 
until you change it. (30, 12) is very large, so on a normal screen the plot may look zoomed or cropped. 
Something like (10, 5) is more typical for a single plot.
'''

#color theme for pie plots
color1 = sns.color_palette('pastel') #color theme for pie plots
color2 = plt.cm.Set2.colors

pie_border = {'edgecolor': 'white'} # border of pie color for easy distinction

plt.subplot(2,3,1)
size = df['gender'].value_counts()
labels = 'Female', 'Male'


plt.pie(size, colors = color1, labels = labels,autopct = '%.1f%%', startangle=90, wedgeprops=pie_border) #pies are made based on size, ie, count
# autopct = '%.1f%%' commonly used. Else '%.2f%%'- more precision OR '%d%%' - whole numbers
# wedgeprops={'edgecolor': 'white'} used for pie slice border
plt.title('Gender', fontsize = 20)
plt.axis('off')



plt.subplot(2,3,2)
size = df['race_ethnicity'].value_counts()
labels = 'Group C', 'Group D','Group B','Group E','Group A'

plt.pie(size, colors = color2,labels = labels,autopct = '%.1f%%', startangle=90, wedgeprops=pie_border)
plt.title('Race/Ethnicity', fontsize = 20)
plt.axis('off')



plt.subplot(2,3,3)
size = df['lunch'].value_counts()
labels = 'Standard', 'Free'

plt.pie(size, colors = color1,labels = labels,autopct = '%.1f%%', startangle=90, wedgeprops=pie_border)
plt.title('Lunch', fontsize = 20)
plt.axis('off')


plt.subplot(2,3,4)
size = df['test_preparation_course'].value_counts()
labels = 'None', 'Completed'

plt.pie(size, colors = color1,labels = labels,autopct = '%.1f%%', startangle=90, wedgeprops=pie_border)
plt.title('Test Course', fontsize = 20)
plt.axis('off')


plt.subplot(2,3,5)
size = df['parental_level_of_education'].value_counts()
labels = 'Some College', "Associate's Degree",'High School','Some High School',"Bachelor's Degree","Master's Degree"


plt.pie(size, colors = color2,labels = labels,autopct = '%.1f%%', startangle=90, wedgeprops=pie_border)
plt.title('Parental Education', fontsize = 20)
plt.axis('off')


plt.tight_layout()
plt.grid()

plt.show()
```

#####  Insights
- Number of Male and Female students is almost equal
- Number students are greatest in Group C
- Number of students who have standard lunch are greater
- Number of students who have not enrolled in any test preparation course is greater
- Number of students whose parental education is "Some College" is greater followed closely by "Associate's Degree"

#### BIVARIATE ANALYSIS (Does gender have an impact on students performance ? ) 

```python
gender_group = df.groupby('gender')['average'].mean()
gender_group
```

```python
gender_group = df.groupby('gender')[['average', 'math_score']].mean()

X = ['Total Average', 'Math Average']
female_scores = gender_group.loc['female'].tolist()   # [average, math_score]
male_scores   = gender_group.loc['male'].tolist()

X_axis = np.arange(len(X))

plt.figure(figsize=(10, 8))
plt.bar(X_axis - 0.2, male_scores, 0.4, label='Male')
plt.bar(X_axis + 0.2, female_scores, 0.4, label='Female')

plt.xticks(X_axis, X)
plt.ylabel("Marks")
plt.title("Total average v/s Math average marks of both the genders", fontweight='bold')
plt.legend()
plt.show()
```

- Females have a better overall score than men.
- males have higher scores in Maths.

#### 4.4.2 RACE/ETHNICITY

```python
f,ax=plt.subplots(1,2,figsize=(20,10))

order = df['race_ethnicity'].value_counts().index
sns.countplot(x='race_ethnicity',data=df,order=order, hue='race_ethnicity', hue_order=order, palette = color1,ax=ax[0],saturation=0.95)
for container in ax[0].containers: #gives value above each bar
    ax[0].bar_label(container,color='black',size=20)
ax[0].set_title('Count by group', fontsize=20)

counts = df['race_ethnicity'].value_counts()
ax[1].pie(counts, labels=counts.index,colors = color1,autopct='%1.1f%%',startangle=90,wedgeprops = pie_border)
ax[1].set_title('Share by group', fontsize=20)

plt.show()
```

#### BIVARIATE ANALYSIS ( Is Race/Ehnicity has any impact on student's performance ? )

```python
f,ax=plt.subplots(1,3,figsize=(20,8))

race_group=df.groupby('race_ethnicity')

#MATH
sns.barplot(x=race_group['math_score'].mean().index,y=race_group['math_score'].mean().values,palette = color1,ax=ax[0])
ax[0].set_title('Math score', size=20)
for container in ax[0].containers:
    ax[0].bar_label(container,color='black',size=15)

#READING
sns.barplot(x=race_group['reading_score'].mean().index,y=race_group['reading_score'].mean().values,palette = color1,ax=ax[1])
ax[1].set_title('Reading score', size=20)
for container in ax[1].containers:
    ax[1].bar_label(container,color='black',size=15)

#WRITING
sns.barplot(x=race_group['writing_score'].mean().index,y=race_group['writing_score'].mean().values,palette = color1,ax=ax[2])
ax[2].set_title('Writing score', size=20)
for container in ax[2].containers:
    ax[2].bar_label(container,color='black',size=15)
```

- Group E students have scored the highest marks. 
- Group A students have scored the lowest marks. 
- Students from a lower Socioeconomic status have a lower avg in all course subjects

- Which type of lunch is most common amoung students ?

#### UNIVARIATE ANALYSIS (Which type of lunch is most common amoung students? )

```python
plt.rcParams['figure.figsize'] = (15, 9)
plt.style.use('seaborn-talk')
sns.countplot(df['lunch'], palette = 'PuBu')
plt.title('Comparison of different types of lunch', fontweight = 30, fontsize = 20)
plt.xlabel('types of lunch')
plt.ylabel('count')
plt.show()
```

#### BIVARIATE ANALYSIS (Impact on student's performance?)

```python
f, ax = plt.subplots(1, 2, figsize=(20, 8))

for a, hue, title in zip(ax,
                         ['test_preparation_course', 'lunch'],
                         ['Students vs test preparation course', 'Students vs lunch type']):
    sns.countplot(data=df, x='parental_level_of_education',
                  hue=hue, palette='bright', saturation=0.95, ax=a)
    a.set_title(title, size=18)
    a.set_xlabel('')
    a.set_ylim(0, 175)
    a.tick_params(axis='x', rotation=30)
    for container in a.containers:
        a.bar_label(container, size=14, padding=2)

plt.tight_layout()
plt.show()
```

- Students who get Standard Lunch tend to perform better than students who got free/reduced lunch
- Master's degree shows low values cause less no. of students have parents with this level of education

#### 4.4.5 TEST PREPARATION COURSE COLUMN 
Impact on student's performance?

#### BIVARIATE ANALYSIS 

```python
f, ax = plt.subplots(1, 3, figsize=(18, 5), sharey=True) #sharey used to normalize y axis vals. across all graphs here

for a, col, title in zip(ax,
                         ['math_score', 'reading_score', 'writing_score'],
                         ['Math', 'Reading', 'Writing']):
    sns.barplot(data=df, x='lunch', y=col,
                hue='test_preparation_course', ax=a)
    a.set_title(title, size=16)
    a.set_xlabel('')
    a.set_ylim(0, 90)
    if a is not ax[-1]:
        a.get_legend().remove()   # keep one legend only

ax[-1].legend(title='Test prep', loc='upper right')
ax[0].set_ylabel('Mean score')
plt.show()
```

- Students who have completed the Test Prepration Course have scores higher in all three categories than those who haven't taken the course
